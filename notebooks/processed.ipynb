```python
df1['platfrom'] = 'Amazon Prime'
df2['platfrom'] = 'Disney Plus'
df3['platfrom'] = 'Hulu'
df4['platfrom'] = 'Netflix'
df1.info()
```

```python
df = pd.concat([df1, df2, df3, df4], ignore_index= True)
df
```

```python
df = df.rename(columns={'platfrom': 'platform'})
```

```python
df
```

```python
# Danh sách các cột dạng văn bản
text_cols = ['director', 'cast', 'country', 'rating', 'type', 'title', 'listed_in', 'description']

# Strip khoảng trắng và điền 'Unknown' cho giá trị NaN/rỗng
for col in text_cols:
    df[col] = df[col].fillna('Unknown').astype(str).str.strip()
    df[col] = df[col].replace({'nan': 'Unknown', '': 'Unknown'})
```

```python
# Chuyển về định dạng datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Điền các ô NaT bằng ngày 01/01 của năm phát hành
missing_dates = df['date_added'].isna()
df.loc[missing_dates, 'date_added'] = pd.to_datetime(
    df.loc[missing_dates, 'release_year'].astype(str) + '-01-01'
)
```

```python
# Tách phần số (duration_int) và phần đơn vị (duration_unit)
df['duration_int'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_unit'] = df['duration'].str.extract(r'([a-zA-Z]+)')
```

```python
# Xóa các dòng bị lặp lại dữ liệu
df = df.drop_duplicates()

# Kiểm tra lại cấu trúc dữ liệu sau khi làm sạch
df.info()
```

```python
# Xóa tất cả ký tự xuống dòng ẩn trong các cột chuỗi
for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].astype(str).str.replace(r'[\r\n\u2028\u2029]', ' ', regex=True)

# Xuất lại file CSV mới
df.to_csv('/home/mortal/python/project/Media_dashboard/data/processed/final_cleaned.csv', index=False)
```
